```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='alcohol', y='cardio', data=df, label='Datos Originales')

# Calculate the mean of the 'cardio' column
mean_cardio = df['cardio'].mean()

# Add a horizontal line at the mean of 'cardio'
plt.axhline(mean_cardio, color='red', linestyle='--', label=f'Promedio Cardio: {mean_cardio:.2f}')

# Calculate y-values for the first new line with intercept 100 and slope 50
slope1 = 50
intercept1 = 100
regression_line_y1 = slope1 * df['alcohol'] + intercept1

# Add the first new line to the plot
plt.plot(df['alcohol'], regression_line_y1, color='green', linestyle='-', label=f'Línea y = {slope1}x + {intercept1}')

# Calculate y-values for the second new line with intercept 300 and slope -50
slope2 = -50
intercept2 = 300
regression_line_y2 = slope2 * df['alcohol'] + intercept2

# Add the second new line to the plot
plt.plot(df['alcohol'], regression_line_y2, color='blue', linestyle='-.', label=f'Línea y = {slope2}x + {intercept2}')

plt.title('Dispersión de Alcohol vs Cardio')
plt.xlabel('Alcohol')
plt.ylabel('Cardio')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

> **Prompt:**
>
> *"generar un modelo de regresión lineal con y = cardio x = alcohol y mostrar la tabla resumen del modelo"*

```python
# Define the linear regression model
model = smf.ols('cardio ~ alcohol', data=df)

# Fit the model
results = model.fit()

# Print the model summary
print(results.summary())
```

> **Prompt (Celda Vacía):**

```python
# Define the linear regression model with 'muertes' as an additional predictor
model_new = smf.ols('cardio ~ alcohol + muertes', data=df)

# Fit the model
results_new = model_new.fit()

# Print the model summary
print(results_new.summary())
```

> **Prompt (Celda Vacía):**

> *"generar un nuevo modelo de regresión agregando hepatic como una variable predictora adicional y mostrar la tabla resumen del modelo"*

```python
# Define the linear regression model with 'alcohol', 'muertes', and 'hepatic' as predictors
model_hepatic = smf.ols('cardio ~ alcohol + muertes + hepatic', data=df)

# Fit the model
results_hepatic = model_hepatic.fit()

# Print the model summary
print(results_hepatic.summary())
```

> **Prompt (Celda Vacía):**
>
> *"generar un nuevo modelo de regresión quitando alcohol como una variable predictora adicional y mostrar la tabla resumen del modelo"*

```python
# Define the linear regression model removing 'alcohol' as a predictor
model_no_alcohol = smf.ols('cardio ~ muertes + hepatic', data=df)

# Fit the model
results_no_alcohol = model_no_alcohol.fit()

# Print the model summary
print(results_no_alcohol.summary())
```

> **Prompt (Celda Vacía):**

> *"generar una tabla comparativa de R-squared, Adj. R-Squared, F-statistic (con su probabilidad), AIC, BIC de los modelos con y = cardio"*

```python
# Extract metrics from each model
model_names = [
    'cardio ~ alcohol',
    'cardio ~ alcohol + muertes',
    'cardio ~ alcohol + muertes + hepatic',
    'cardio ~ muertes + hepatic' # Added the new model
]
models = [results, results_new, results_hepatic, results_no_alcohol] # Added results_no_alcohol

data = []
for i, model_result in enumerate(models):
    row = {
        'Model': model_names[i],
        'R-squared': model_result.rsquared,
        'Adj. R-squared': model_result.rsquared_adj,
        'F-statistic': model_result.fvalue,
        'Prob (F-statistic)': model_result.f_pvalue,
        'AIC': model_result.aic,
        'BIC': model_result.bic
    }
    data.append(row)

# Create DataFrame
comparison_df = pd.DataFrame(data)

display(comparison_df.round(4))
```
