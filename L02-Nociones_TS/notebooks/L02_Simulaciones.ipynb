```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# **Simulaciones Ruido Blanco**

```python
numPuntos = 200 # Número de puntos por simulación
numSim = 100 # Número de simulaciones
```

```python
# Crear un arreglo de simulaciones de dimensiones (numPuntos, numSim)
wSim = np.random.randn(numPuntos, numSim)

# Convertir el arreglo np en df
df_wSim = pd.DataFrame(wSim)

display(df_wSim.tail())
```

```python
df = df_wSim.copy()
nombre = 'Ruido Blanco'

# Graficar todas las simulaciones
plt.figure(figsize=(12, 6))

for col in df.columns:
    plt.plot(df[col], alpha=0.5, linewidth=1)

plt.title(str(numSim) + ' Simulaciones de ' + nombre)
plt.xlabel('Periodo (Puntos)')
plt.ylabel('x_t')
plt.grid(True)
plt.show()
```

# **Simulaciones Caminata Aleatoria**

```python
# Crear un nuevo DataFrame con la suma acumulativa de cada columna de wSimDF
df_walkSim = df_wSim.cumsum()

df_walkSim.tail()
```

```python
df = df_walkSim.copy()
nombre = 'Caminata Aleatoria'

# Graficar todas las simulaciones
plt.figure(figsize=(12, 6))

for col in df.columns:
    plt.plot(df[col], alpha=0.5, linewidth=1)

plt.title(str(numSim) + ' Simulaciones de ' + nombre)
plt.xlabel('Periodo (Puntos)')
plt.ylabel('x_t')
plt.grid(True)
plt.show()
```

# **Simulaciones Caminata Aleatoria con Drift**

```python
drift = 0.3
df_walkDriftSim = (df_wSim + drift).cumsum()
df_walkDriftSim.head()
```

```python
df = df_walkDriftSim.copy()
nombre = 'Caminata Aleatoria con Drift ' + str(drift)

# Graficar todas las simulaciones
plt.figure(figsize=(12, 6))

for col in df.columns:
    plt.plot(df[col], alpha=0.5, linewidth=1)

plt.title(str(numSim) + ' Simulaciones de ' + nombre)
plt.xlabel('Periodo (Puntos)')
plt.ylabel('x_t')
plt.grid(True)
plt.show()
```
